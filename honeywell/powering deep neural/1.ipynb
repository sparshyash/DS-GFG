Language models like GPT (Generative Pre-trained Transformer) have revolutionized the field
of natural language processing by demonstrating remarkable abilities in generating
human-quality text.

A sampling strategy, such as greedy decoding (selecting the most probable word) or
more sophisticated methods like temperature sampling (introducing randomness), is used to
choose the next word

Forward propagation is the “prediction step.”
Forward propagation is the process of sending input data through the network to get an
output, moving from the input layer to the output layer. Backward propagation (or
backpropagation) is the process of sending the error, calculated after the forward pass, back
through the network to adjust its internal parameters (weights and biases) to minimize that error
