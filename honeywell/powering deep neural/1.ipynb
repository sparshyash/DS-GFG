Forward propagation is the “prediction step.”
Forward propagation is the process of sending input data through the network to get an
output, moving from the input layer to the output layer. Backward propagation (or
backpropagation) is the process of sending the error, calculated after the forward pass, back
through the network to adjust its internal parameters (weights and biases) to minimize that error


Backpropagation
What is Backward Propagation?
Backward propagation is the “learning step.”
Backpropagation, short for Backward Propagation of Errors, is a key algorithm used to train
neural networks by minimizing the difference between predicted and actual outputs. It works by
propagating errors backward through the network, using the chain rule of calculus to compute
gradients and then iteratively updating the weights and biases. Combined with optimization
techniques like gradient descent, backpropagation enables the model to reduce loss across
epochs and effectively learn complex patterns from data.
After the network makes a guess:
1. It checks how wrong the guess was (error).
2. The error moves backward through the network.
3. Each neuron adjusts its weights slightly to reduce future mistake

Activation Functions
Activation functions are crucial components of neural networks that introduce non-linearity into
the output of a neuron. Without activation functions, a neural network would simply be a linear
regression model, regardless of its depth. Activation functions enable the network to learn
complex, non-linear relationships in the data, making them essential for tackling real-world
problems

Examples: ReLU, Sigmoid, Tanh, Softmax.



Loss Functions
A loss function, also known as a cost function, quantifies the difference between the model's
predictions and the actual target values. The goal of training is to minimize this loss function,
effectively making the model's predictions as close as possible to the true values. The choice of
loss function depends on the type of task being performe
