```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

np.random.seed(42) # for reproducibility
n_samples = 1000
data = {
    'distance': np.random.uniform(1, 10, n_samples), # Distance in kilometers
    'num_items': np.random.randint(1, 10, n_samples), # Number of items in the order
    'restaurant_rating': np.random.uniform(3, 5, n_samples), # Restaurant rating (3-5)
    'delivery_time': 0 # Placeholder for delivery time
}
df = pd.DataFrame(data)
df['delivery_time'] = (
    5 + # Base time
    df['distance'] * 3 + # Distance impact
    df['num_items'] * 1.5 - # Number of items impact
    df['restaurant_rating'] * 0.5 + # Restaurant rating impact
    np.random.normal(0, 5, n_samples) # Add some noise
)
df['delivery_time'] = df['delivery_time'].clip(lower=0)
print(df.head())
```

This code snippet generates a Pandas DataFrame with 1000 rows. The features are:
● distance: A random number between 1 and 10, representing the delivery distance in kilometers.
● num_items: A random integer between 1 and 10, representing the number of items in the order.
● restaurant_rating: A random number between 3 and 5, representing the restaurant's rating.
● delivery_time: The target variable, calculated based on the other features with added noise to
simulate real-world variability
